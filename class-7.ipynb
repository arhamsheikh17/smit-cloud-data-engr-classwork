```python
table=int(input("Enter number"))
times= int(input("Enter how many times u want to print "))
for temp in range(times):
    print(f"{table}*{temp+1} = {table*(temp+1)}")
```

```python
num=11
for x in [2,3,4,5]:
    for y in range(1,num):
        print(f"{x}*{y}={x*y}")
    num-=1   
    print("------------------")
```

## list and dictionary comprehension 

```python
l = [i**2 for i in range(1,8,2)]
l ## return list with operation of power of 2
```

```python
d = {ii:ii**2 for ii in range(5)}
d
```

```python
result ={"ali":80, "hamza":67,"asad": 70,"umair":77}
s = {name:marks for name, marks in result.items() if marks > 70}
s
```
