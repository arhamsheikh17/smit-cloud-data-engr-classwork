```python
## Functions recursion (discuss in class with code )
## while loops
##    while condition: (syntax)
num = int(input("Enter your factorial number"))          # change this to any positive integer
result = 1        # factorial starts from 1
i = 1
while i <= num:
    
    result *= i   # same as result = result * i
    i += 1        # move to the next number

print(f"Factorial of {num} is {result}")

```
