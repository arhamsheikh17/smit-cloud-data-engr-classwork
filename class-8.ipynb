## Function 

```python
def calculator():
    a=int(input("Enter first number"))
    b= int(input("Enter second number "))
    c= input("which operation you want to perform +,_,*,/")
    if c == '+':
        print(a+b)
    elif  c == '_':
        print(a-b)
    elif c =='*':
        print(a*b)
    elif c =='/':
        print(a/b)    
    else:
        print("Invalid function")  
calculator()          

```
