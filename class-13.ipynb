```python
## Example
class Student:
    def __init__(self, name, grade):
        self.name = name
        self.__grade = grade   # private attribute

    # Public method
    def show_name(self):
        print(f"Name: {self.name}")

    # Protected method
    def _show_greeting(self):
        print(f"Hello, {self.name}! (from protected method)")

    # Private method
    def __show_grade(self):
        print(f"Grade: {self.__grade}")

    # Public gateway method
    def show_all(self, choice):
        if choice == "name":
            self.show_name()
        elif choice == "greet":
            self._show_greeting()
        elif choice == "grade":
            self.__show_grade()
        else:
            print("Invalid choice!")


# Using the class
s1 = Student("Ali", "A")

s1.show_all("name")   # calls public method
s1.show_all("greet")  # calls protected method
s1.show_all("grade")  # calls private method
```
