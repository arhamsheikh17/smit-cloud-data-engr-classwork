## shallow / deep copy

```python
#shallow copy diff adress per rkhta hy 
## deep copy same adress ko use krti hy , kisi ek ko bhi change krengy dosra bhi change hojaiga 
##sth sth jb ky shallow ma nh hita kisi ek ky change sy dosry ma change
```

```python
## deep example
```

```python
import copy

original = [23, 3,4,5,6]
deep_copied = copy.deepcopy(original)

# Modify the copy
deep_copied[0] = 99

print("Original:", original)      # [[1, 2], [3, 4]]
print("Deep Copy:", deep_copied)  # [[99, 2], [3, 4]]
```

```python
id(original)
id(deep_copied)
```
