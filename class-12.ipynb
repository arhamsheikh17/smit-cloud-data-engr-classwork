```python
## Task
class father():

    def __init__(self,  name, designtion):
        self.name = name
        self.designtion = designtion

    def canScold(self):
        print(f"Mr {self.name} has right to scold bcz he {self.designtion} ")    


class mother():

    def __init__(self,  prob_solver):
        self.prob_solver = prob_solver
        
class child(father, mother):
    

    def __init__(self, name, designtion, prob_solver, gender, age):
        father.__init__(self, name, designtion)
        mother.__init__(self, prob_solver)
        self.gender = gender
        self.age = age

    def canDrive(self):
        print(f"Child gender is {self.gender} and age is {self.age}")


child_obj = child("arham", "lwyer", "yes", "male",18)      

child_obj.canScold()
child_obj.canDrive()

print(child_obj.name, child_obj.age, child_obj.prob_solver)
child_obj.gender
```

```python
## overridding (constructor , method) , overloading 
## jb inherit hotii hy tuu same method agr child ma bh hogaa tu woh parent method ko xonsider nh krega child ko consider krega 
# or os method ko use kr ky output degaa 
## Overloading (@overload sy just create a showcase of method , actual functionality humm hum @overload ky block ma nahi dalty 
# we create it in other block where we perform functionality ) --> appprox same as abstraction 
from typing import overload
class adder():
    @overload
    def add(self, x: int, y: int) -> int:
        ...
    @overload
    def add(self, x: float, y: float)  -> float:
        ...
    @overload
    def add(self, x: str, y: str)  -> str:
        ...

    def add(self, x, y):
        if isinstance(x, int) and isinstance(y, int):
            return x+y
        elif isinstance(x, str) and isinstance(y, str):
            return x+y    
        elif isinstance(x, float) and isinstance(y, float):
            return x+y 
        else:
            print("Type error")

obj = adder()
obj.add(2,4)
obj.add(2.22,4.55)
obj.add("3","7")
obj.add("3",4)
## Abstraction....

```
