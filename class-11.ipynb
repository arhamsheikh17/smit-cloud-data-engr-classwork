## Lambda function

```python
## a func has no name , never been called before or after...., no return keyword 
## syntax ---> lambda inputs:expression 
## map , filter , reduce , lambda func (self study), aggregation function(sum , max, min)
## any --> any one cond true , all means all conditions should true...
f = lambda n: 1 if n==1 else n*f(n-1)
f(4)
## lambda arguments: expression
# lambda → the keyword.
# arguments → like function parameters (can be 0 or more).
# expression → a single expression that gets evaluated and returned.

square = lambda x: x * x
print(square(5))

## With two arguments

add = lambda a, b: a + b
print(add(3, 7))


```

```python
class Temperature:
    @staticmethod
    def c_to_f(c):
        return (c * 9/5) + 32

    @staticmethod
    def f_to_c(f):
        return (f - 32) * 5/9

print(Temperature.c_to_f(0))   # 32.0
print(Temperature.f_to_c(212)) # 100.0
```
